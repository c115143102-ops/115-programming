```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0

# =================================================================
#  【核心參數抗干擾調校區】— 完美破解黑夜中烏鴉晚跳版
# =================================================================
# 寬度稍微拉開到 30 像素（220~240），防止黑夜切換時網頁掉幀、瞬移穿透！
DETECTION_X_START = 220 
DETECTION_X_END   = 240        
        
# 高度維持 425，精準抓中空烏鴉的肚子與地面仙人掌
DETECTION_Y_START = 425   
DETECTION_Y_END   = 460         
        
# 【終極核心修正】從原本的 0.12 調小到 0.09！
# 縮短空中停留時間，強迫小恐龍快起快落，拋物線距離自然就會變短，防範飛過頭死掉！
AIRTIME   = 0.09                 

# 起跳冷卻維持 0.05
COOLDOWN  = 0.05       
# =================================================================
   
print("【🌓 縮短跳躍距離特調版 🌓】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)
 
last_jump_time = 0
is_jumping = False

while True:
    current_time = time.time()
    
    # 1. 抓取精準的動態防禦條
    bbox = (DETECTION_X_START, DETECTION_Y_START, DETECTION_X_END, DETECTION_Y_END)
    im = ImageGrab.grab(bbox=bbox)
    
    # 2. 自動抓取當下左上角的像素作為「即時背景」
    bg_pixel = im.getpixel((0, 0))
    colors = im.getcolors()
    
    has_obstacle = False
    if colors:
        for count, color in colors:
            # 排除背景色，且必須有超過 2 個像素點才算障礙物（抗雜訊）
            if color != bg_pixel and count > 2:
                
                try:
                    # 計算這組顏色跟當前背景的 RGB 差距
                    rdiff = abs(color[0] - bg_pixel[0])
                    gdiff = abs(color[1] - bg_pixel[1])
                    bdiff = abs(color[2] - bg_pixel[2])
                    
                    # 【終極核心修正】將原本的 > 10 降到 > 3！
                    # 這樣在白天切黑夜、網頁還在漸層變灰的時候，烏鴉一現身就能瞬間被抓到，絕不晚跳！
                    if rdiff > 3 or gdiff > 3 or bdiff > 3:
                        has_obstacle = True
                        break
                except (TypeError, IndexError):
                    # 安全機制：若抓到非 RGB 格式的單色值直接比對絕對值
                    if abs(color - bg_pixel) > 3:
                        has_obstacle = True
                        break
                
    # 3. 狀態機控制：空中時間到，立刻解鎖狀態
    if is_jumping and (current_time - last_jump_time > AIRTIME):
        pyg.keyUp('space') # 強制釋放跳躍鍵
        is_jumping = False
        
    # 4. 觸發防禦機制
    if has_obstacle:
        if current_time - last_jump_time > COOLDOWN:
            # 只洗掉下蹲鍵（down），絕對不碰 space，根治雙擊大跳的 Bug！
            pyg.keyUp('down')
            
            # 觸發最純粹、乾淨的光速點放起跳
            pyg.keyDown('space')
            pyg.keyUp('space')
            
            last_jump_time = time.time()
            is_jumping = True
```

```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 建議先手動測試你的偵測區域 (X=200, Y=420, 寬=20, 高=40)
detection_box = (200, 420, 220, 460)

print("【⚔️ 仙人掌與鳥·完美分流版 ⚔️】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0

while True:
    current_time = time.time()
    
    # 保持微秒級超快截圖
    im = ImageGrab.grab(bbox=detection_box)
    width, height = im.size
    triggered = False
    
    # 這裡我們恢復用座標讀取，才能知道東西到底是在「空中」還是「地面」
    for x in range(width):
        for y in range(height):
            c = im.getpixel((x, y))
            
            # 日夜通用排除法
            if c != (255, 255, 255) and c != (32, 33, 36):
                
                # =================【 關鍵核心：高低空分流防禦 】=================
                # 如果異物出現在偵測區的前 15 像素（上半部），代表這是高空鳥！
                if y < 15:
                    # 遇到高空鳥，絕對不往上跳！直接使出「閃電下蹲」滑過去
                    pyg.keyDown('down')
                    time.sleep(0.12)  # 下蹲時間
                    pyg.keyUp('down')
                    
                # 如果異物出現在下半部，代表是地面仙人掌或低空鳥
                else:
                    # 只有距離上一次跳躍超過落地時間，才觸發中跳
                    if current_time - last_jump_time > 0.22:
                        pyg.keyDown('space')
                        time.sleep(0.08)  # 保持你最愛的最穩中跳高度
                        pyg.keyUp('space')
                        last_jump_time = time.time()
                        
                triggered = True
                break
        if triggered:
            break
```
