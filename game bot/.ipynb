```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 建議先手動測試你的偵測區域 (X=200, Y=420, 寬=20, 高=40)
detection_box = (200, 420, 220, 460)

print("【⚔️ 仙人掌與鳥·完美分流版 ⚔️】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0

while True:
    current_time = time.time()
    
    # 保持微秒級超快截圖
    im = ImageGrab.grab(bbox=detection_box)
    width, height = im.size
    triggered = False
    
    # 這裡我們恢復用座標讀取，才能知道東西到底是在「空中」還是「地面」
    for x in range(width):
        for y in range(height):
            c = im.getpixel((x, y))
            
            # 日夜通用排除法
            if c != (255, 255, 255) and c != (32, 33, 36):
                
                # =================【 關鍵核心：高低空分流防禦 】=================
                # 如果異物出現在偵測區的前 15 像素（上半部），代表這是高空鳥！
                if y < 15:
                    # 遇到高空鳥，絕對不往上跳！直接使出「閃電下蹲」滑過去
                    pyg.keyDown('down')
                    time.sleep(0.12)  # 下蹲時間
                    pyg.keyUp('down')
                    
                # 如果異物出現在下半部，代表是地面仙人掌或低空鳥
                else:
                    # 只有距離上一次跳躍超過落地時間，才觸發中跳
                    if current_time - last_jump_time > 0.22:
                        pyg.keyDown('space')
                        time.sleep(0.08)  # 保持你最愛的最穩中跳高度
                        pyg.keyUp('space')
                        last_jump_time = time.time()
                        
                triggered = True
                break
        if triggered:
            break
```

```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 建議先手動測試你的偵測區域 (X=200, Y=420, 寬=20, 高=40)
detection_box = (200, 420, 220, 460)

print("【🛡️ 零衝突·完美相容分流版 🛡️】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0

while True:
    current_time = time.time()
    
    # 保持微秒級超快截圖
    im = ImageGrab.grab(bbox=detection_box)
    width, height = im.size
    triggered = False
    
    for x in range(width):
        for y in range(height):
            c = im.getpixel((x, y))
            
            # 日夜通用排除法：只要有東西進入這個 20x40 的區塊
            if c != (255, 255, 255) and c != (32, 33, 36):
                
                # =================【 防 誤 判 空 中 分 流 】=================
                if y < 15 and (c == (83, 83, 83) or c == (172, 172, 172)):
                    # 遇到高空鳥，改用 press 點一下放開，不要用 keyDown 卡住迴圈
                    pyg.press('down')
                    triggered = True
                    break
                    
                # =================【 防 卡 鍵 地 面 分 流 】=================
                elif y >= 15:
                    # 只有距離上一次跳躍超過落地時間，才觸發中跳
                    if current_time - last_jump_time > 0.22:
                        # 【關鍵核心】起跳前，強制把可能按著的下蹲鍵拔起來，保證絕對會跳！
                        pyg.keyUp('down') 
                        
                        pyg.keyDown('space')
                        time.sleep(0.08)  # 保持你最愛的最穩中跳高度
                        pyg.keyUp('space')
                        last_jump_time = time.time()
                        
                triggered = True
                break
        if triggered:
            break
```
