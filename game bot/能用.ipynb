```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0

# 這是你設定的黃金防禦區域：(X=215, Y=420, 寬=20, 高=40)
detection_box = (200, 420, 220, 460)

# =================【 滯 空 時 間 極 限 調 整 區 】=================
# 原本是 0.1 秒。現在改為 0.07 秒！
# 數字越小，在空中飄的時間越短、落地越快。如果會撞到仙人掌頂端再微調加大（如 0.08 或 0.09）
AIR_TIME = 0.8
# =================================================================

print(f"【⚡ 閃電下墜 {AIR_TIME}s 極限版 ⚡】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0

while True:
    current_time = time.time()
    
    # 微秒級超快截圖
    im = ImageGrab.grab(bbox=detection_box)
    width, height = im.size
    triggered = False
    
    for x in range(width):
        for y in range(height):
            c = im.getpixel((x, y))
            
            # 日夜通用排除法
            if c != (255, 255, 255) and c != (32, 33, 36):
                
                # =================【 防 誤 判 空 中 鳥 類 】=================
                if y < 15 and (c == (83, 83, 83) or c == (172, 172, 172)):
                    pyg.press('down')
                    triggered = True
                    break
                    
                # =================【 防 禦 地 面 仙 人 掌 】=================
                elif y >= 15:
                    if current_time - last_jump_time > 0.22:
                        pyg.keyUp('down') # 起跳前保險解鎖下蹲
                        
                        # 1. 觸發起跳
                        pyg.keyDown('space')
                        pyg.keyUp('space')
                        
                        # 2. 控制極限滯空時間
                        time.sleep(AIR_TIME) 
                        
                        # 3. 立刻執行閃電下墜，強迫牠砸回地面！
                        pyg.keyDown('down')
                        pyg.keyUp('down')
                        
                        last_jump_time = time.time()
                        
                triggered = True
                break
        if triggered:
            break
```
