```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0

# =================================================================
#  【核心參數抗干擾調校區】— 請用 mouseInfo() 量到的數字直接填入
# =================================================================
# 1. 根據 mouseInfo() 看到的仙人掌/中空鳥位置，填入防線 X 軸
DETECTION_X_START = 200     # 👈 起跳防線起點
DETECTION_X_END   = 225     # 👈 保持比 START 多 25 像素，防後期瞬移穿透
        
# 2. 根據 mouseInfo() 看到的鳥類高度過濾：
# DETECTION_Y_START 必須「大於」高空鳥最底端，且「小於」中空鳥的肚子！
DETECTION_Y_START = 425     # 👈 抬頭高度（防空分水嶺）
DETECTION_Y_END   = 460     # 👈 地面高度（踩在土地上的高度）
        
# 3. 動作控制（解除腳麻、完美拋物線）
AIRTIME   = 0.14            # 👈 落地時間：太早砸地踩頭死就調大；太慢落地撞下一個就調小
COOLDOWN  = 0.15            # 👈 起跳冷卻：連續四個仙人掌跳不起來就調小（例如 0.11）
# =================================================================
   
print("【🎯 專業級 mouseInfo 對號入座版】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)
 
last_jump_time = 0
is_jumping = False

while True:
    current_time = time.time()
    
    # 1. 抓取精準動態防禦條
    bbox = (DETECTION_X_START, DETECTION_Y_START, DETECTION_X_END, DETECTION_Y_END)
    im = ImageGrab.grab(bbox=bbox)
    
    # 2. 自動抓取當下左上角的像素作為「即時背景」
    bg_pixel = im.getpixel((0, 0))
    colors = im.getcolors()
    
    has_obstacle = False
    if colors:
        for count, color in colors:
            # 排除背景色，且排除少數像素的雜訊
            if color != bg_pixel and count > 2:
                
                try:
                    # 計算這組顏色跟當前背景的 RGB 差距，日夜交替免疫
                    rdiff = abs(color[0] - bg_pixel[0])
                    gdiff = abs(color[1] - bg_pixel[1])
                    bdiff = abs(color[2] - bg_pixel[2])
                    
                    if rdiff > 10 or gdiff > 10 or bdiff > 10:
                        has_obstacle = True
                        break
                except (TypeError, IndexError):
                    if abs(color - bg_pixel) > 10:
                        has_obstacle = True
                        break
                
    # 3. 狀態機控制：空中時間到，立刻強迫砸回地面
    if is_jumping and (current_time - last_jump_time > AIRTIME):
        pyg.keyUp('space') # 強制釋放空白鍵，避免底層卡鍵
        pyg.keyDown('down')
        pyg.keyUp('down')
        is_jumping = False
        
    # 4. 觸發防禦機制
    if has_obstacle:
        if current_time - last_jump_time > COOLDOWN:
            # 物理洗掉下蹲狀態，防卡死
            pyg.keyUp('down')
            pyg.keyUp('down')
            
            # 起跳
            pyg.keyDown('space')
            pyg.keyUp('space')
            
            last_jump_time = time.time()
            is_jumping = True
```
