```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0

# 黃金防禦區域：(X=220, Y=420, 寬=20, 高=40)
detection_box = (235, 420, 255, 460)

# =================================================================
#  【核心參數調整區】
# =================================================================
AIR_TIME = 0.12      # 👈 【落下時間】原本 0.18，調小（例如 0.12）會更快砸回地面！
COOLDOWN = 0.15      # 👈 【起跳冷卻】搭配落下時間縮短，冷卻也要同步調小，防連跳卡死
# =================================================================

print(f"【⚡ 落下時間 {AIR_TIME}s 極速調校版 ⚡】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0
is_jumping = False

while True:
    current_time = time.time()
    
    # 1. 微秒級超快截圖
    im = ImageGrab.grab(bbox=detection_box)
    bg_color = im.getpixel((0, 0))
    colors = im.getcolors()
    
    has_obstacle = False
    for count, color in colors:
        if color != bg_color:
            has_obstacle = True
            break
            
    # 2. 狀態機控制：空中時間到，立刻強迫砸回地面
    if is_jumping and (current_time - last_jump_time > AIR_TIME):
        pyg.keyDown('down')
        pyg.keyUp('down')   # 點按即放，絕不殘留
        is_jumping = False
        
    # 3. 觸發防禦機制
    if has_obstacle:
        if current_time - last_jump_time > COOLDOWN:
            # 強制洗掉底層下蹲狀態，確保跳躍 100% 成功
            pyg.keyUp('down')
            pyg.keyUp('down')
            
            # 觸發跳躍
            pyg.keyDown('space')
            pyg.keyUp('space')
            
            last_jump_time = time.time()
            is_jumping = True
```

```python
import pyautogui as pyg
import time
from PIL import ImageGrab

# 核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0

# =================================================================
#  【核心參數抗干擾調校區】
# =================================================================
DETECTION_X_START = 230 # 看遠點
DETECTION_X_END   = 245 # 看近點       
  
DETECTION_Y_START = 420   
DETECTION_Y_END   = 460         
  
AIR_TIME = 0.08     # 落地時間
COOLDOWN = 0.18      # 起跳冷卻
# =  ================================================================
  
print("【🌓 日夜切換免疫版 🌓】已啟動...")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)

last_jump_time = 0
is_jumping = False

while True:
    current_time = time.time()
    
    # 1. 抓取極精細的地面防禦條
    bbox = (DETECTION_X_START, DETECTION_Y_START, DETECTION_X_END, DETECTION_Y_END)
    im = ImageGrab.grab(bbox=bbox)
    
    # 2. 自動抓取當下左上角的像素作為「即時背景」
    bg_pixel = im.getpixel((0, 0))
    colors = im.getcolors()
    
    has_obstacle = False
    if colors:
        for count, color in colors:
            # 【關鍵修正】：不只比對單一顏色，而是去算它與即時背景的「色差」
            # 白天切夜間時，背景色 bg_pixel 會即時更新，所以不會誤判！
            if color != bg_pixel and count > 2:
                
                # 計算這組顏色跟當前背景的 RGB 差距
                # 如果差距太小，代表只是日夜切換時的過渡期雜訊，不予理會
                rdiff = abs(color[0] - bg_pixel[0])
                gdiff = abs(color[1] - bg_pixel[1])
                bdiff = abs(color[2] - bg_pixel[2])
                
                if rdiff > 10 or gdiff > 10 or bdiff > 10:
                    has_obstacle = True
                    break
                
    # 3. 狀態機控制：空中時間到，立刻強迫砸回地面
    if is_jumping and (current_time - last_jump_time > AIR_TIME):
        pyg.keyDown('down')
        pyg.keyUp('down')
        is_jumping = False
        
    # 4. 觸發防禦機制
    if has_obstacle:
        if current_time - last_jump_time > COOLDOWN:
            pyg.keyUp('down')
            pyg.keyUp('down')
            
            pyg.keyDown('space')
            pyg.keyUp('space')
            
            last_jump_time = time.time()
            is_jumping = True
```
