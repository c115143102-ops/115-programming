```python

import time
import pyautogui as pyg

# ========================================================
# ⚙️ 請根據你的螢幕，調整這兩個核心起點（非常重要）
# ========================================================
START_X = 697  # 小恐龍嘴尖前方空地的 X 座標
DETECT_Y = 407  # 剛好切過仙人掌身體的 Y 座標

CACTUS_COLOR = (83, 83, 83)  # 仙人掌的灰色


def run_dino_bot():
  print('請在 3 秒內切換到小恐龍遊戲視窗...')
  time.sleep(3)
  pyg.press('space')  # 開始遊戲
  print('動態外掛已啟動！')

  # 記錄遊戲開始的時間
  start_time = time.time()

  while True:
    # 1. 計算遊戲已經進行了幾秒
    elapsed_time = time.time() - start_time

    # 2. 【核心機制】基本看 40 像素遠，每過 1 秒，就多看 2.5 像素遠
    # 這樣隨著遊戲加速，小恐龍也會自動越看越遠
    current_scan_width = int(40 + (elapsed_time * 2.5))

    # 限制最大偵測距離（例如 250 像素），避免看太遠看到下一個障礙物導致連續亂跳
    if current_scan_width > 250:
      current_scan_width = 250

    # 3. 快速截取這一條動態長度的橫線
    line_box = pyg.screenshot(
        region=(START_X, DETECT_Y, current_scan_width, 1)
    )
    pixels = line_box.getdata()

    # 4. 檢查是否有仙人掌闖入這個隨時間拉長的雷達區
    if CACTUS_COLOR in pixels:
      pyg.press('space')

      # 當速度變極快時，甚至可以讓牠在空中直接按「下」快速落地（可選）
      # pyg.press('down')

      # 跳躍後的短暫冷卻，隨速度變快，冷卻時間要縮減
      if elapsed_time < 30:
        time.sleep(0.15)
      else:
        time.sleep(0.05)  # 後期速度極快，必須立刻準備下一次跳躍


if __name__ == '__main__':
  run_dino_bot()
```

```python
import time
from PIL import Image, ImageGrab
import pyautogui

# 修正按鍵控制，確保按完一定會放開，絕對不卡死、不抽搐


def click_action(key):
  if key == 'up':
    pyautogui.press('space')  # 按一下就放開，防止連續抽搐起跳
  elif key == 'down':
    pyautogui.keyDown('down')
    time.sleep(0.15)  # 蹲下 0.15 秒躲中間烏鴉
    pyautogui.keyUp('down')  # 💡 必須放開！


def isCollision(data):
  # 在這塊只有障礙物的乾淨圖片中，自動判斷白天或黑夜
  bg_color = data[0, 0]
  is_night = bg_color < 128

  # --- 1. 偵測半空中（中、高空烏鴉）---
  # 在切出來的局部小圖中，Y軸 0~40 是空中
  for x in range(0, 90):
    for y in range(0, 40):
      pixel_val = data[x, y]
      if (not is_night and pixel_val < 100) or (is_night and pixel_val > 150):
        click_action('down')  # 發現烏鴉，蹲下！
        return

  # --- 2. 偵測地面（仙人掌、低空鳥）---
  # 在切出來的局部小圖中，Y軸 65~80 是地面
  for x in range(0, 90):
    for y in range(65, 80):
      pixel_val = data[x, y]
      if (not is_night and pixel_val < 100) or (is_night and pixel_val > 150):
        click_action('up')  # 發現仙人掌，跳！
        return


if __name__ == '__main__':
  print('=== 👑 局部雷達必勝版（徹底解決抽搐） ===')
  print('請在 5 秒內將小恐龍遊戲網頁切換到前台並放到最大...')
  time.sleep(5)
  pyautogui.press('space')  # 開始遊戲

  while True:
    # 💡 終極修改：不要抓全螢幕！
    # 我們用 bbox 直接在螢幕上切出一塊「寬 90、高 80」的純淨偵測方塊
    # 橫向從 650 到 740（剛好在小恐龍右手邊前方，絕對碰不到恐龍自己的身體！）
    # 縱向從 350 到 430（涵蓋空中到地面）
    image = ImageGrab.grab(bbox=(650, 350, 740, 430)).convert('L')

    data = image.load()
    isCollision(data)
```
